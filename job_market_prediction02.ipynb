```python
# Create a separate copy
jobs_df = df.copy()

# Remove rows without job titles
jobs_df = jobs_df.dropna(subset=['title'])

# Clean text columns
jobs_df['title'] = jobs_df['title'].astype(str).str.strip()

jobs_df['country'] = (
    jobs_df['country']
    .fillna('Unknown')
    .astype(str)
    .str.strip()
)

# Convert date column
jobs_df['published_date'] = pd.to_datetime(
    jobs_df['published_date'],
    errors='coerce',
    utc=True
)

# Convert hourly rates into numeric values
for col in ['hourly_low', 'hourly_high', 'budget']:
    jobs_df[col] = pd.to_numeric(
        jobs_df[col],
        errors='coerce'
    )

# Remove duplicate job records
jobs_df = jobs_df.drop_duplicates(
    subset=['title', 'link']
)

# Create hourly midpoint
jobs_df['hourly_midpoint'] = np.where(
    jobs_df['is_hourly'].eq(True),
    (jobs_df['hourly_low'] + jobs_df['hourly_high']) / 2,
    np.nan
)

# Create recommendation text
jobs_df['recommendation_text'] = (
    jobs_df['title'].fillna('') + ' ' +
    jobs_df['country'].fillna('')
)

# Reset index
jobs_df = jobs_df.reset_index(drop=True)

# Inspect cleaned dataset
print("Cleaned dataset shape:", jobs_df.shape)

print("\nMissing values:")
display(jobs_df.isnull().sum())

print("\nSample records:")
display(jobs_df.head())

print("\nJob type distribution:")
display(jobs_df['is_hourly'].value_counts(dropna=False))
```

```python
# Make a working copy
jobs_df = jobs_df.copy()

# Remove records without a job title
jobs_df = jobs_df.dropna(subset=['title'])

# Clean text columns
jobs_df['title'] = jobs_df['title'].astype(str).str.strip()
jobs_df['job_category'] = jobs_df['job_category'].astype(str).str.strip()

# Handle missing countries
jobs_df['country'] = jobs_df['country'].fillna('Unknown').astype(str).str.strip()

# Convert date column
jobs_df['published_date'] = pd.to_datetime(
    jobs_df['published_date'], errors='coerce', utc=True
)

# Convert numeric columns safely
numeric_cols = ['hourly_low', 'hourly_high', 'budget']

for col in numeric_cols:
    jobs_df[col] = pd.to_numeric(jobs_df[col], errors='coerce')

# Normalize hourly/fixed-price indicator
jobs_df['is_hourly'] = (
    jobs_df['is_hourly']
    .astype(str)
    .str.strip()
    .str.lower()
    .map({'true': True, 'false': False})
)

# Calculate hourly midpoint where both rates are available
jobs_df['hourly_midpoint'] = (
    jobs_df['hourly_low'] + jobs_df['hourly_high']
) / 2

# Create a readable job type
jobs_df['job_type'] = jobs_df['is_hourly'].map({
    True: 'Hourly',
    False: 'Fixed-price'
}).fillna('Unknown')

# Create text used by the recommendation engine
jobs_df['recommendation_text'] = (
    jobs_df['title'] + ' ' + jobs_df['job_category']
)

# Remove duplicate postings only when a non-empty link is duplicated
jobs_df = jobs_df.drop_duplicates(
    subset=['link'],
    keep='first'
)

# Reset index
jobs_df = jobs_df.reset_index(drop=True)

# Check the result
print("Prepared Dataset Shape:", jobs_df.shape)
print("\nMissing Values:")
display(jobs_df.isnull().sum())

print("\nJob Type Distribution:")
display(jobs_df['job_type'].value_counts())

print("\nSample Prepared Records:")
display(
    jobs_df[
        ['title', 'job_category', 'country', 'job_type',
         'hourly_midpoint', 'budget', 'recommendation_text']
    ].head(10)
)
```

```python
# Build the TF-IDF feature matrix

from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF vectorizer
tfidf = TfidfVectorizer(
    stop_words='english',
    ngram_range=(1, 2),
    max_features=50000,
    min_df=2,
    max_df=0.95
)

# Convert job descriptions/titles into TF-IDF vectors
job_vectors = tfidf.fit_transform(
    jobs_df['recommendation_text']
)

# Display results
print("TF-IDF Matrix Shape:", job_vectors.shape)
print("Total Vocabulary Size:", len(tfidf.vocabulary_))
print("Matrix Type:", type(job_vectors))
print("Matrix Sparsity:", round(
    100 * (1 - job_vectors.nnz /
          (job_vectors.shape[0] * job_vectors.shape[1])), 2
), "%")
```

```python
# Create the job recommendation function
from sklearn.metrics.pairwise import cosine_similarity

def recommend_jobs(
    profile,
    country=None,
    job_type=None,
    min_rate=None,
    max_rate=None,
    top_n=10
):
    # Validate profile
    if not profile or not profile.strip():
        raise ValueError("Please enter a job title or skills.")

    # Convert user profile into TF-IDF vector
    profile_vector = tfidf.transform([profile])

    # Calculate similarity scores
    similarity_scores = cosine_similarity(
        profile_vector,
        job_vectors
    ).flatten()

    # Add scores to a copy of the dataset
    results = jobs_df.copy()
    results['similarity_score'] = similarity_scores

    # Keep jobs with a text match
    results = results[results['similarity_score'] > 0]

    # Optional country filter
    if country:
        results = results[
            results['country'].str.lower() == country.strip().lower()
        ]

    # Optional job type filter
    if job_type:
        results = results[
            results['job_type'].str.lower() == job_type.strip().lower()
        ]

    # Optional rate or budget filters
    if min_rate is not None or max_rate is not None:

        if job_type and job_type.lower() == 'fixed-price':
            amount = results['budget']
        elif job_type and job_type.lower() == 'hourly':
            amount = results['hourly_midpoint']
        else:
            raise ValueError(
                "Please select Hourly or Fixed-price when using rate filters."
            )

        if min_rate is not None:
            results = results[amount >= min_rate]

        if max_rate is not None:
            results = results[amount <= max_rate]

    # Sort by similarity
    results = results.sort_values(
        by='similarity_score',
        ascending=False
    )

    # Return top recommendations
    output_cols = [
        'title',
        'job_category',
        'country',
        'job_type',
        'hourly_low',
        'hourly_high',
        'hourly_midpoint',
        'budget',
        'similarity_score',
        'link'
    ]

    return results[output_cols].head(top_n).reset_index(drop=True)
```

```python
# Test recommendation engine

recommendations = recommend_jobs(
    profile="Data Analyst SQL Python Power BI Excel data visualization",
    country="India",
    top_n=10
)

display(recommendations)
```

```python
# Create the interactive job recommendation UI

# Get available countries and categories
countries = sorted(jobs_df['country'].dropna().unique().tolist())
categories = sorted(jobs_df['job_category'].dropna().unique().tolist())

def job_recommendation_ui(
    profile,
    country,
    job_type,
    category,
    top_n
):
    try:
        # Convert empty selections to None
        country_filter = None if country == "All Countries" else country
        job_type_filter = None if job_type == "All Types" else job_type
        category_filter = None if category == "All Categories" else category

        # Generate recommendations
        result = recommend_jobs(
            profile=profile,
            country=country_filter,
            job_type=job_type_filter,
            category=category_filter,
            top_n=int(top_n)
        )

        if result.empty:
            return pd.DataFrame(
                columns=[
                    'title', 'job_category', 'country',
                    'job_type', 'similarity_score', 'link'
                ]
            )

        # Select display columns
        result = result[
            [
                'title',
                'job_category',
                'country',
                'job_type',
                'hourly_midpoint',
                'budget',
                'similarity_score',
                'link'
            ]
        ].copy()

        result['similarity_score'] = result[
            'similarity_score'
        ].round(4)

        return result

    except Exception as e:
        raise gr.Error(str(e))


# Build Gradio interface
with gr.Blocks(title="Job Recommendation Engine") as demo:

    gr.Markdown("# Job Recommendation Engine")
    gr.Markdown(
        "Find relevant job postings based on your skills and preferences."
    )

    profile = gr.Textbox(
        label="Enter Job Title or Skills",
        placeholder="Example: Data Analyst, Python, SQL, Power BI, Excel",
        lines=3
    )

    with gr.Row():
        country = gr.Dropdown(
            choices=["All Countries"] + countries,
            value="All Countries",
            label="Preferred Country"
        )

        job_type = gr.Dropdown(
            choices=["All Types", "Hourly", "Fixed-price"],
            value="All Types",
            label="Job Type"
        )

    category = gr.Dropdown(
        choices=["All Categories"] + categories,
        value="All Categories",
        label="Preferred Job Category"
    )

    top_n = gr.Slider(
        minimum=5,
        maximum=20,
        value=10,
        step=1,
        label="Number of Recommendations"
    )

    recommend_button = gr.Button(
        "Find Recommended Jobs",
        variant="primary"
    )

    output = gr.Dataframe(
        label="Recommended Jobs",
        interactive=False
    )

    recommend_button.click(
        fn=job_recommendation_ui,
        inputs=[
            profile,
            country,
            job_type,
            category,
            top_n
        ],
        outputs=output
    )

# Launch the application
demo.launch(share=True)
```
